The dataset was loaded into a Pandas DataFrame using `pd.read_csv()`. After loading, I examined:

## Step 2: Handle Missing Values

I checked the number of missing values in each column using `isna().sum()`.

The missing values were handled as follows:

- **Age:** Missing values were filled with the **mean age**, since Age is a numerical feature and replacing missing values with the mean preserves most of the information.
- **Cabin:** The entire column was dropped because it contained a large number of missing values, making it less useful for analysis.
- **Embarked:** Missing values were filled with the **most frequent value (mode)** because Embarked is a categorical feature.

After these operations, the dataset no longer contains missing values in the remaining columns.

```python
df.isna().sum()
```

```python
df['Age']=df['Age'].fillna(df['Age'].mean())
df=df.drop('Cabin',axis=1)
df['Embarked']=df['Embarked'].fillna(df['Embarked'].mode()[0])
df.head()
```

```python
df.isna().sum()
```

## Step 3: Filter the Dataset

The dataset was filtered to include only passengers whose **Age is greater than 30**.

This filtering creates a subset containing older passengers, allowing analysis of this specific age group. The resulting dataset contains fewer rows than the original, as younger passengers were excluded.

```python
df=df[df['Age']>30]
print(df.shape)
df.head()
```

## Step 4: Grouping and Aggregation

The data was grouped by the **Sex** column using `groupby()`. For each group, the following statistics were computed:

- **Maximum Age**
- **Average Fare**
- **Most Common Embarkation Port**

These aggregated statistics summarize the characteristics of male and female passengers. Comparing these values helps identify differences between the two groups in terms of age, ticket fare, and the most common embarkation location.

```python
df.groupby("Sex").agg({"Age":"max","Fare":'mean','Embarked':lambda x:x.mode()[0]})
```

### Interpretation

The data is grouped by passenger sex.

- The maximum age is higher for males than for females.
- The average fare paid by females is higher than that of males.
- The most common embarkation port differs (or is the same) between the two groups.

These statistics show differences in age, ticket price, and embarkation patterns between male and female passengers.
