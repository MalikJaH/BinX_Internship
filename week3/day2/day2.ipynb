```python
import matplotlib.pylab as plt
import numpy as np
from IPython.display import Image, display
```

```python
weeks = [1,2,3,4,5]
cars_sales = [2,4,5,4,5]
plt.scatter(weeks,cars_sales)
plt.xticks(range(0,6,1))
plt.yticks(range(0,6,1))
```

We need to know where to draw the best line that describe the data
,first calculate the mean for x and y.
X the weeks and y the cars_sales

```python
x_mean = np.mean(weeks)
y_mean = np.mean(cars_sales)
plt.scatter([x_mean],[y_mean])
plt.axvline(x=x_mean, color='orange')
plt.axhline(y=y_mean, color='orange')
```

after we know the point that the line will go through it, we need to know the angle

```python
# calculate first x-x_mean
weeks_mean = [x - x_mean.item() for x in weeks ]
# calculate  y-y_mean
cars_sales_mean = [x - y_mean.item() for x in cars_sales ]

#but the idea is now the negative removes the positve numbers so need to use another formula which is (x - x_mean)^2
weeks_mean_seq = [x * x for x in weeks_mean]
weeks_mean_seq_sum = np.sum(weeks_mean_seq).item()


print('weeks_mean: ',weeks_mean)
print('weeks_mean_seq: ',weeks_mean_seq)
print('weeks_mean_seq_sum: ',weeks_mean_seq_sum)

print('cars_sales_mean: ',cars_sales_mean)
```

We need now to knew the area and we can calculate it with (x-x_mean)(y-y_mean)

```python
display(Image(filename='assets/area.png'))
display(Image(filename='assets/calculations.png'))
```

```python
areas= [a * b for a, b in zip(cars_sales_mean, weeks_mean)]
areas_sum = np.sum(areas).item()
print('areas: ',areas)
print('areas_sum: ',areas_sum)
```

Now calculate the y using the formula y = mx+c

```python
display(Image(filename='assets/slope.png'))
display(Image(filename='assets/c.png'))

```

```python
m = areas_sum / weeks_mean_seq_sum
c = y_mean - (m*x_mean)
print('m: ',m)
print('c: ',c)
```

### **Residual** is the distance between the actual point and the predicted point
To get the **best fit** of the line we need to change **m** and **c** togther
