### Step 2: Compute the correlation matrix and visualize it as an annotated heatmap.

```python
corr = df.corr(numeric_only=True)
sns.heatmap(corr,
            annot=True,
            cmap="coolwarm")
```

#### Step 3: Identify the two or three strongest relationships and state what they might mean for a future model.

I noticed two relationships that appear:

- **Fare and Pclass:** There is a strong negative relationship between these two features. Passengers in higher classes (lower `Pclass` values). This relationship may be useful for a future model because both features capture information about a passenger's status, which is known to be related to survival in the Titanic dataset.

- **SibSp and Parch:** These features show a positive relationship because they both describe family members traveling with a passenger. People traveling with siblings or spouses often also traveled with parents or children. Combining these features into a new feature such as `FamilySize` could provide a more informative representation of family size for a machine learning model.
