## Step 1

I simulated 10,000 fair coin flips using NumPy.

The proportion of heads is very close to 0.5.

```python
np.random.seed(42)
coin_flips= np.random.randint(0,2,10000)
heads= coin_flips.sum()
tails = len(coin_flips) - heads
print("Heads: ",heads)
print("Tails: ",tails)
print("Propapilty of heads: ",(heads/len(coin_flips)))
```

## Step 3

The idea is we have 10,000 trails to get a card from a shuffled bag
we need to calculate the probability of getting Lose card after having a Win card
theoreticaly:

P (Second Lose | First Win) = 2 / 4 = 0.5

The simulated probability is very close to 0.5, confirming the theoretical result.

```python
trials = 100000

success = 0
condition = 0

bag = ['W', 'L', 'L', 'W', 'W']
for _ in range(trials):
    np.random.shuffle(bag)

    first = bag[0]
    second = bag[1]

    if first == 'W':
        condition += 1

        if second == 'L':
            success += 1

print("Simulated Conditional Probability:", success / condition)
```
