```python
import pandas as pd 
import matplotlib.pyplot as plt
import numpy as np
```

---

## Step 1: Load and Clean the Dataset

The Titanic dataset was loaded using Pandas.

The following preprocessing steps were performed:

- Filled missing values in the **Age** column using the mean age.
- Removed the **Cabin** column because it contains a large number of missing values.
- Filled missing values in the **Embarked** column using the most frequent value (mode).
- Verified the dataset structure using `df.info()`.

---

```python
df = pd.read_csv("../day4/train.csv")
df['Age']=df['Age'].fillna(df['Age'].mean())
df=df.drop('Cabin',axis=1)
df['Embarked']=df['Embarked'].fillna(df['Embarked'].mode()[0])
df.info()
```

---

## Step 2: Compute a Summary Statistic

Using NumPy, I calculated the average age of the passengers.

This summary statistic provides a quick overview of the dataset and was also used to replace missing values in the **Age** column.

---

```python
mean_age = np.mean(df["Age"])
print(mean_age)
```

---

## Visualization 1: Age vs. Fare (Scatter Plot)

This scatter plot shows the relationship between passengers' ages and their ticket fares.

The plot helps identify whether there is any correlation between age and fare. In this dataset, passengers of different ages paid a wide range of fares, suggesting that age alone does not strongly determine the ticket price.

---

```python
plt.scatter(df['Age'], df['Fare'])
plt.title("Fare Per Person Relationship")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```


## Visualization 2: Age Distribution (Histogram)

The histogram illustrates the distribution of passengers' ages.

It shows that most passengers were young adults, while fewer passengers were very young or elderly. This visualization makes it easy to understand how the ages are distributed across the dataset.

---

```python
plt.hist(df['Age'])
plt.title("Age Destribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

## Visualization 3: Age by Sex (Bar Chart)

This bar chart compares passenger ages by gender.

Because every passenger is plotted individually, the chart becomes crowded and is not the best choice for this type of data. A bar chart is generally more suitable for comparing aggregated values, such as the average age of males and females.

---

```python
plt.bar(df["Sex"],df["Age"])
plt.title('Comparison of Age Values with Sex')
plt.xlabel("Sex")
plt.ylabel("Age")
plt.show()
```

## Visualization 4: Subplots

The final figure combines two visualizations into a single window using Matplotlib subplots.

The left subplot displays the bar chart, while the right subplot shows the age distribution histogram. Subplots make it easier to compare multiple visualizations side by side.

```python
fig, axes= plt.subplots(1,2,figsize=(10,4))
axes[0].bar(df["Sex"],df["Age"])
axes[0].set_title('Comparison of Age Values with Sex')
axes[0].set_xlabel("Sex")
axes[0].set_ylabel("Age")
axes[1].hist(df['Age'])
axes[1].set_title("Age Destribution")
axes[1].set_xlabel("Age")
axes[1].set_ylabel("Frequency")
plt.show()
```
