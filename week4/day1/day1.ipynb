# Day 1 - Week 4: Train / Validation / Test Splits

## Main Idea

In Week 3, we mainly used:

**Train → Test**

That is enough when we train a model once and only evaluate it once.

However, when we start trying different models or hyperparameters, we need another dataset for making those development decisions.

The professional workflow becomes:

**Train → Validation → Test**

Each set has a different job.

## 1. Train, Validation, and Test

| Dataset | Purpose | When it is used |
|---|---|---|
| **Training Set** | The model learns from this data | During `.fit()` |
| **Validation Set** | Used to compare settings and tune the model | During development |
| **Test Set** | Final honest evaluation | Once, at the very end |

### Important Rule

The **test set must stay untouched during tuning**.

If we keep checking the test score and changing our model based on it, the test set starts influencing our decisions.

Then it is no longer a truly unseen final evaluation set.

## 2. Why a Validation Set Is Needed

Suppose we want to test different values for `n_neighbors` in KNN:

- `k = 3`
- `k = 5`
- `k = 7`
- `k = 9`

If we compare all of them directly on the **test set**, then the test set is helping us choose the model.

That means we are indirectly tuning the model to the test data.

Instead, we should:

1. Train each model using the **training set**.
2. Compare the models using the **validation set**.
3. Select the best setting.
4. Use the **test set only once** for the final score.
