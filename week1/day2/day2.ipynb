## Step 1
this function job is to recive **list of nubmers** and then return **mean**, **max** and **min** as a dict 

```python
def stats(numbers):
    return {"mean":sum(numbers)/len(numbers),"max":max(numbers),"min":min(numbers)} 
print(stats([1,2,3,4,5]))
```

## Step 2
this fucntion job is to recive a list of numbers and keep only the **even** ones

```python
def even(numbers):
    return [x for x in numbers if x%2==0]
print(even([1,2,3,4,5]))
```
